1. Introducción y objeto del informe
El presente documento contiene el análisis del informe interactivo "10.000 Empresas Más Grandes del País 2026", publicado por la Superintendencia de Sociedades de Colombia a través de un dashboard de Microsoft Power BI de acceso público. El informe consolida y clasifica a las diez mil (10.000) sociedades del sector real no financiero con mayores ingresos operacionales del país, con base en la información financiera reportada a los distintos entes de supervisión (Superintendencia de Sociedades, Superintendencia Financiera, Superintendencia de Vigilancia y Seguridad Privada, Superintendencia Nacional de Salud, Superintendencia de Transporte y Superintendencia de Servicios Públicos Domiciliarios).
El objetivo de este análisis es sintetizar la información visualizada en el tablero (metodología, indicadores financieros agregados, rankings por ingresos y ganancias, distribución sectorial y regional, insolvencia empresarial y sostenibilidad) y presentar una lectura crítica de los principales hallazgos.
2. Advertencia y licencia de uso de los datos
El informe advierte que los datos publicados son de carácter público. En virtud de la Ley 1712 de 2014 (Ley de Transparencia y del Derecho de Acceso a la Información Pública Nacional), los datos pueden usarse, aprovecharse o transformarse libremente, siempre que no se desnaturalice su sentido y se cite como fuente a la Superintendencia de Sociedades (www.supersociedades.gov.co). El informe está avalado por la Superintendente de Sociedades (e), el Superintendente Delegado para Asuntos Económicos y Societarios, la Dirección de Información Empresarial y los equipos de Estudios Económicos y Contables.
3. Metodología del estudio
El dashboard documenta de forma explícita diez criterios metodológicos, que constituyen la base de validez técnica del ranking:
Fuentes de información: seis entidades de supervisión estatal (Supersociedades, Superfinanciera, Supervigilancia, Supersalud, Supertransporte y Superservicios).
Universo de estudio: entidades del sector real no financiero.
Variables financieras recibidas: activos, pasivos, patrimonio, ingresos operacionales y ganancias/pérdidas.
Corte de la información: 31 de diciembre de 2025, con fecha de recepción oportuna al 25 de mayo de 2026.
Estandarización contable: cifras bajo Normas Internacionales de Información Financiera (NIIF).
Criterio de tamaño: ingresos operacionales (actividades ordinarias, otros ingresos y participación en subsidiarias, asociadas y negocios conjuntos bajo el método de participación).
Naturaleza de los estados financieros: individuales y separados; no se emplean estados consolidados.
Nivel de depuración: la información se presenta tal como fue reportada por las sociedades, sin validación de fondo por parte de la Superintendencia (salvo la homologación de unidad monetaria a miles de pesos).
Comparabilidad interanual: el comparativo 2024–2025 se realiza sobre cifras reportadas a corte de cada año, sin ajustes ni retransmisiones.
Lectura crítica: la metodología es transparente y coherente para un ejercicio de ranking, pero tiene dos limitaciones relevantes que deben tenerse presentes al interpretar los resultados: (i) al no validarse la información de fondo, persiste el riesgo de inconsistencias o errores de reporte por parte de las propias sociedades, y (ii) el uso de estados financieros individuales (no consolidados) puede subestimar o sobreestimar el verdadero tamaño económico de grupos empresariales con múltiples sociedades vinculadas.

10. Sostenibilidad e informes ante la Superintendencia
El tablero identifica 1.559 sociedades que han presentado informe de sostenibilidad ante la Superintendencia de Sociedades. Adicionalmente, permite consultar la participación de sociedades en índices y reconocimientos de sostenibilidad como el Anuario Global de Sostenibilidad, el Dow Jones Best-in-Class World Index y el reconocimiento Emisor IR (ejemplo destacado: Chevron), con fuente en S&P Dow Jones Indices, Merco Colombia y la Bolsa de Valores de Colombia.
Lectura crítica: 1.559 sociedades sobre un universo de 10.000 (15,6 %) equivale a una adopción todavía minoritaria, aunque creciente, de prácticas de reporte de sostenibilidad (ESG) entre las grandes empresas colombianas. Esto es consistente con la tendencia regional, en la que la sostenibilidad corporativa formal —más allá de la responsabilidad social tradicional— aún se concentra en compañías grandes, multinacionales o emisoras de valores sujetas a estándares internacionales.
11. Conclusiones del informe original
El propio dashboard presenta las siguientes conclusiones consolidadas para la vigencia 2025:
Los activos consolidados de las 10.000 empresas más grandes alcanzaron $2.560,4 billones en 2025, con una variación anual de 4,2 %.
La estructura de financiación cerró en 54 % pasivos y 46 % patrimonio.
Los ingresos operacionales agregados ascendieron a $1.853,8 billones, con una variación anual de 5,03 %.
Los ingresos operacionales alcanzaron una magnitud cercana al PIB nominal estimado de Colombia para 2025.
El crecimiento de los ingresos estuvo respaldado principalmente por comercio, servicios y manufactura.
Las utilidades netas alcanzaron $138,6 billones en 2025, con un crecimiento de 13,09 % frente al año anterior.
Las pérdidas reportadas disminuyeron en $7,3 billones y el número de empresas con pérdidas se redujo en 216 sociedades frente a 2024.
12. Análisis integral y conclusiones propias
12.1 Fortalezas del ejercicio
Transparencia metodológica: el informe documenta con precisión sus fuentes, alcance, período de corte y limitaciones, lo cual facilita su interpretación y uso por terceros.
Accesibilidad e interactividad: al estar construido en Power BI y publicado bajo licencia abierta (Ley 1712 de 2014), el dashboard permite a cualquier usuario filtrar por macrosector, región, vigencia o sociedad específica, lo que amplía enormemente su utilidad frente a un informe estático.
Cobertura amplia: 10.000 sociedades del sector real no financiero constituyen una muestra representativa de la actividad empresarial formal del país.
12.2 Riesgos e interpretaciones a considerar
Al usar estados financieros individuales (no consolidados), el ranking puede fragmentar el verdadero tamaño de grandes grupos empresariales que operan a través de múltiples sociedades vinculadas, subestimando su peso relativo.
La información no fue validada de fondo por la Superintendencia, por lo que puede contener inconsistencias originadas en el reporte de las propias sociedades.
La comparación de ingresos agregados con el PIB nominal, aunque útil como referencia de magnitud, no debe interpretarse como una medición homogénea, pues corresponde a conceptos contables distintos (ingresos brutos vs. valor agregado).
12.3 Balance general
En conjunto, el informe "10.000 Empresas Más Grandes del País 2026" retrata un tejido empresarial colombiano que, en 2025, mostró crecimiento moderado en ingresos (5,03 %) acompañado de un crecimiento más que proporcional en utilidades (13,09 %), una estructura de financiamiento apalancada pero solvente, una alta concentración sectorial en hidrocarburos, comercio y servicios públicos, y una reducción de la insolvencia empresarial medida en número de procesos y activos comprometidos frente al año anterior. La adopción de prácticas de sostenibilidad, aunque en expansión, sigue siendo minoritaria dentro del universo analizado, lo que representa tanto un reto como una oportunidad de mejora para el sector corporativo del país.
